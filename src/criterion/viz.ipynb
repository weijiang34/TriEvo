```python
import numpy as np
import os
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
def plot_loss(loss_dir):
    loss_path = os.path.join(loss_dir, 'loss.npy')
    perplexity_path = os.path.join(loss_dir, 'perplexity.npy')

    loss = np.load(loss_path)
    perplexity = np.load(perplexity_path)
    
    fig = plt.figure(figsize=(10, 5))
    ax1 = fig.add_subplot(2, 1, 1)
    ax1.plot(loss, label='Loss')
    ax1.set_xlabel('Token Index')
    ax1.set_ylabel('Loss')
    ax1.set_title('Per-token Loss')
    ax1.legend()
    ax1.grid()

    ax2 = fig.add_subplot(2, 1, 2)
    ax2.plot(perplexity, label='Perplexity')
    ax2.set_xlabel('Token Index')
    ax2.set_ylabel('Perplexity')
    ax2.set_title('Per-token Perplexity')
    ax2.legend()
    ax2.grid()

    # fig.tight_layout()
    # fig.savefig(os.path.join(loss_dir, 'loss_perplexity_plot.png'))

plot_loss('/srv/scratch/z3543429/LLM/TriEvo/dataset/bacteria/NC_000964.3/phage_seqs/phage_pbsx.fasta.perbase')
```
