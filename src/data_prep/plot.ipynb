```python
import pandas as pd
import os
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
CLUSTER_DIR = "/srv/scratch/z3543429/LLM/TriEvo/dataset/refseq_viral/cluster"
ANI_LIST = [95, 90, 85, 75]
ani = 95
cluster = pd.read_csv(os.path.join(CLUSTER_DIR, f"filtered_clusters_ani_{ani}.tsv"), sep='\t', header=None, names=['rep_id', 'all_ids'])

def plot_cluster_sizes(cluster_df, ani):
    cluster_df['cluster_size'] = cluster_df['all_ids'].apply(lambda x: len(x.split(',')))
    plt.figure(figsize=(10, 6))
    sns.histplot(cluster_df['cluster_size'], bins=30, kde=True)
    plt.title(f'Cluster Size Distribution at ANI {ani}%')
    plt.xlabel('Cluster Size')
    plt.ylabel('Frequency')

plot_cluster_sizes(cluster, ani)
```
